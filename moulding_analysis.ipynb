```python
master_df = pd.read_excel(
    "MASTER DATA OF MOULDING.xlsx",
    header=1
)
```

```python
master_df.head()
```

```python
print(master_df.columns)
```

```python
master_df.columns = master_df.iloc[0]
```

```python
master_df = master_df[1:]
```

```python
master_df = master_df.reset_index(drop=True)
```

```python
master_df.head()
```

```python
print(master_df.columns)
```

```python
master_df.info()
```

```python
numeric_columns = [
    'Charge Weight (Grams)',
    'shift output moulding',
    'SHIFT RATE',
    'MOTOR HP',
    'HEATER WATT',
    'MATERIAL RATE KG',
    'DEFLASHING RATE',
    'shift output of 8 hrs',
    'INS AND PACKING',
    'shift output for packing',
    'SCHEDULE'
]

for col in numeric_columns:
    master_df[col] = pd.to_numeric(
        master_df[col],
        errors='coerce'
    )
```

```python
master_df.info()
```

```python
master_df['RM Required KG'] = (
    master_df['SCHEDULE'] *
    master_df['Charge Weight (Grams)']
) / 1000
```

```python
master_df[
    [
        'Part No',
        'SCHEDULE',
        'Charge Weight (Grams)',
        'RM Required KG'
    ]
].head()
```

```python

master_df['Required Shifts'] = (
    master_df['SCHEDULE'] /
    master_df['shift output moulding']
)
```

```python
master_df[
    [
        'Part No',
        'SCHEDULE',
        'shift output moulding',
        'Required Shifts'
    ]
].head()
```

```python
master_df['Available Shifts'] = (
    master_df['machine tonnage']
    .map({
        '100T': 300,
        '150T': 250
    })
)
```

```python
master_df[
    [
        'Part No',
        'machine tonnage',
        'Available Shifts'
    ]
].head()
```

```python
print(master_df.columns)
```

```python
for col in master_df.columns:
    print(col)
```

```python
master_df[
    [
        'Part No',
        'Mould No',
        'SCHEDULE'
    ]
].head(10)
```

```python
mould_schedule = (
    master_df.groupby('Mould No')['SCHEDULE']
    .sum()
    .reset_index()
)
```

```python
mould_schedule.head(10)
```

```python
mould_output = (
    master_df.groupby('Mould No')['shift output moulding']
    .first()
    .reset_index()
)
```

```python
mould_summary = pd.merge(
    mould_schedule,
    mould_output,
    on='Mould No'
)
```

```python
mould_summary['Required Shifts'] = (
    mould_summary['SCHEDULE'] /
    mould_summary['shift output moulding']
)
```

```python
mould_summary.head(10)
```

```python
mould_machine = (
    master_df.groupby('Mould No')['machine tonnage']
    .first()
    .reset_index()
)
```

```python
mould_summary = pd.merge(
    mould_summary,
    mould_machine,
    on='Mould No'
)
```

```python
mould_summary['Available Shifts'] = (
    mould_summary['machine tonnage']
    .map({
        '100T': 300,
        '150T': 250
    })
)
```

```python
mould_summary['Utilisation %'] = (
    mould_summary['Required Shifts'] /
    mould_summary['Available Shifts']
) * 100
```

```python
mould_summary.head(10)
```

```python
total_rm = master_df['RM Required KG'].sum()

print("Total RM Required KG :", total_rm)
```

```python
total_shifts = mould_summary['Required Shifts'].sum()

print("Total Required Shifts :", total_shifts)
```

```python
avg_utilisation = mould_summary['Utilisation %'].mean()

print("Average Machine Utilisation % :", avg_utilisation)
```

```python
import plotly.express as px
```

```python
fig = px.bar(
    mould_summary,
    x='Mould No',
    y='Utilisation %',
    color='machine tonnage',
    title='Machine Utilisation by Mould'
)

fig.show()
```

```python
master_df['RM Cost'] = (
    master_df['RM Required KG'] *
    master_df['MATERIAL RATE KG']
)
```

```python
master_df[
    [
        'Part No',
        'RM Required KG',
        'MATERIAL RATE KG',
        'RM Cost'
    ]
].head(10)
```

```python
master_df['MATERIAL RATE KG'] = (
    master_df['MATERIAL RATE KG']
    .fillna(method='ffill')
)
```

```python
master_df['MATERIAL RATE KG'] = (
    master_df['MATERIAL RATE KG']
    .ffill()
)
```

```python
master_df['RM Cost'] = (
    master_df['RM Required KG'] *
    master_df['MATERIAL RATE KG']
)
```

```python
master_df[
    [
        'Part No',
        'RM Required KG',
        'MATERIAL RATE KG',
        'RM Cost'
    ]
].head(10)
```

```python
total_rm_cost = master_df['RM Cost'].sum()

print("Total RM Cost :", total_rm_cost)
```

```python
rm_summary = (
    master_df.groupby('Raw Material Grade')
    ['RM Required KG']
    .sum()
    .reset_index()
)
```

```python
fig = px.bar(
    rm_summary,
    x='Raw Material Grade',
    y='RM Required KG',
    title='Raw Material Requirement by Grade'
)

fig.show()
```

```python
with pd.ExcelWriter('Moulding_Planning_Output.xlsx') as writer:
    
    master_df.to_excel(
        writer,
        sheet_name='Master Planning',
        index=False
    )
    
    mould_summary.to_excel(
        writer,
        sheet_name='Mould Summary',
        index=False
    )
    
    rm_summary.to_excel(
        writer,
        sheet_name='RM Summary',
        index=False
    )
```
